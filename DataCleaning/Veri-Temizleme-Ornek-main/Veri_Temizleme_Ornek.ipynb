**Veri Dönüştürme**  

Bazı satırları uygun bir formata dönüştürmemiz gerekiyor.

```python
def extract_number_from_string(val):
    if isinstance(val, str) and val.startswith('[') and val.endswith(']'):
        try:
            return int(val.strip('[]'))
        except:
            return val
    return val

data['room'] = data['room'].apply(extract_number_from_string)
data['livingRoom'] = data['livingRoom'].apply(extract_number_from_string)
```

```python
import json

def extract_name(val):
    # Handle stringified JSON (if needed)
    if isinstance(val, str):
        try:
            val = json.loads(val)
        except:
            return val

    # Handle list of dicts (like in 'areas')
    if isinstance(val, list) and len(val) > 0 and isinstance(val[0], dict):
        return val[0].get('name')

    # Handle plain dicts (like 'district', 'county', etc.)
    if isinstance(val, dict):
        # Check for 'typeName' in subCategory structure
        if 'typeName' in val:
            return val.get('typeName')  # For subCategory, return typeName
        return val.get('name')

    return val

columns_to_transform = ['areas', 'district', 'county', 'city', 'residence','heating','fuel','build','buildState','usage','credit','barter','housingComplex','subCategory']

for col in columns_to_transform:
    data[col] = data[col].apply(extract_name)

# Display to verify
print(data[columns_to_transform].head())
```

```python
def extract_net_sqm(val):
    if isinstance(val, str):
        try:
            val = json.loads(val)
        except:
            return None
    if isinstance(val, dict):
        return val.get('netSqm')
    return None

def extract_gross_sqm(val):
    if isinstance(val, str):
        try:
            val = json.loads(val)
        except:
            return None
    if isinstance(val, dict):
        gross = val.get('grossSqm')
        if isinstance(gross, list) and len(gross) > 0:
            return gross[0]
    return None

# Create the new columns
data['netSqm'] = data['sqm'].apply(extract_net_sqm)
data['grossSqm'] = data['sqm'].apply(extract_gross_sqm)

# Drop original 'sqm' ,no longer needed
data.drop(columns='sqm', inplace=True)

# Preview
print(data[['netSqm', 'grossSqm']].head())
```

```python
def extract_floor_count(val):
    if isinstance(val, str):
        try:
            val = json.loads(val)
        except:
            return None
    if isinstance(val, dict):
        return val.get('count')
    return None

# Update the floor column with just the count
data['floor'] = data['floor'].apply(extract_floor_count)
```

```python
import numpy as np

# 1. residence
residence_map = {
    # canonical names
    'Daire':          'Daire',
    'Tripleks':       'Tripleks',
    'Müstakil Ev':    'Müstakil Ev',
    'MüstakilEv':     'Müstakil Ev',
    'Dubleks':        'Dubleks',
    'Bahçe Katı':     'Bahçe Katı',
    'BahçeKatı':      'Bahçe Katı',
    'Bahçe Dubleksi': 'Bahçe Dubleksi',
    'Çatı Dubleksi':  'Çatı Dubleksi',
    'Ters Dubleks':   'Ters Dubleks',
    'Ara Kat Dubleks':'Ara Kat Dubleks',
    'İkiz Ev':        'İkiz Ev',
    # weird/corrupted → Unknown
    None:             'Unknown',
    'None':           'Unknown',
    # everything else (rare) → Other
}
data['residence'] = (
    data['residence']
      .replace(residence_map)
      .fillna('Other')
      .where(lambda s: s.isin([
         'Daire','Tripleks','Müstakil Ev','Dubleks',
         'Bahçe Katı','Bahçe Dubleksi','Çatı Dubleksi',
         'Ara Kat Dubleks','Ters Dubleks','İkiz Ev'
      ]), other='Other')
)

# 2. build
build_map = {
    'Betonarme':              'Betonarme',
    'Reinforced Concrete':    'Betonarme',
    'Bet tonarme':            'Betonarme',
    'Taş Bina':               'Taş Bina',
    'TaşBina':                'Taş Bina',
    'Kagir':                  'Kagir',
    'Ahşap':                  'Ahşap',
    'Çelik':                  'Çelik',
    'Yığma':                  'Yığma',
    'Prefabrik':              'Prefabrik',
    None:                     'Unknown',
    'None':                   'Unknown',
}
data['build'] = data['build'].replace(build_map).fillna('Unknown')

# 3. buildState → New / Second‑hand / Under Construction / Unknown
build_state_map = {
    'Sıfır':               'Sıfır',
    'Sifir':               'Sıfır',
    '0':                   'Sıfır',
    'İkinci El':           'İkinci El',
    'IkinciEl':            'İkinci El',
    'İkinciEl':            'İkinci El',
    'Ikinci El':            'İkinci El',
    'Second Hand':         'İkinci El',
    'Kiracılı':            'İkinci El',
    'Yapım Aşamasında':    'Sıfır',
    None:                  'Unknown',
    'None':                'Unknown',
}
data['buildState'] = data['buildState'].replace(build_state_map).fillna('Unknown')

# 4. usage → Vacant / Owner‑occupied / Rented / Unknown
usage_map = {
    'Boş':               'Boş',
    'Bos':               'Boş',
    'Bo�':               'Boş',
    'Empty':             'Boş',
    'Mülk Sahibi':       'Mülk Sahibi',
    'MülkSahibi':        'Mülk Sahibi',
    'M�lk Sahibi':       'Mülk Sahibi',
    'Owner':             'Mülk Sahibi',
    'Kiracılı':          'Kiracılı',
    'Kiracili':          'Kiracılı',
    'Tenanted':          'Kiracılı',
    'Belirtilmemiş':     'Unknown',
    None:                'Unknown',
    'None':              'Unknown',
}
data['usage'] = data['usage'].replace(usage_map).fillna('Unknown')

# 5. credit → Available / Not Available / Unknown
credit_map = {
    'Uygun':            'Uygun',
    'Available':        'Uygun',
    'Uygun değil':      'Uygun değil',
    'Uygundeğil':       'Uygun değil',
    'Uygun degil':      'Uygun değil',
    'Uygun de�il':      'Uygun değil',
    'Not Available':    'Uygun değil',
    'Notavail':         'Uygun değil',
    'Bilinmiyor':       'Unknown',
    None:               'Unknown',
    'None':             'Unknown',
}
data['credit'] = data['credit'].replace(credit_map).fillna('Unknown')

# 6. barter → Allowed / Not Allowed / Unknown
barter_map = {
    'Yapılır':    'Yapilir',
    'Yapilir':    'Yapilir',
    'Makes':      'Yapilir',
    'Yap�l�r':    'Yapilir',
    'Yapılmaz':   'Yapilmaz',
    'Yap�lmaz':   'Yapilmaz',
    'Yapilmaz':   'Yapilmaz',
    'Not Done':   'Yapilmaz',
    None:         'Unknown',
    'None':       'Unknown',
}
data['barter'] = data['barter'].replace(barter_map).fillna('Unknown')

# 7. landRegisterName → Condominium / Floor Easement / No Title / Land / Other / Unknown
land_map = {
    'Kat Mülkiyeti':     'Kat Mülkiyeti',
    'KatMülkiyeti':      'Kat Mülkiyeti',
    'Kat M�lkiyeti':     'Kat Mülkiyeti',
    'Condominium':       'Kat Mülkiyeti',
    'Kat İrtifakı':      'Kat İrtifakı',
    'Katİrtifakı':       'Kat İrtifakı',
    'Kat Irtifaki':      'Kat İrtifakı',
    'Kat �rtifak�':      'Kat İrtifakı',
    'Floor Easement':    'Kat İrtifakı',
    'Tapu Yok':          'Tapu Yok',
    'TapuYok':           'Tapu Yok',
    'Arsa':              'Arsa',
    None:                'Unknown',
    'None':              'Unknown',
}
data['landRegisterName'] = data['landRegisterName'].replace(land_map).fillna('Unknown')


# 8. furnished → 1 / 0
# fill NaN as False, then convert
data['furnished'] = data['furnished'].fillna(False).map({True:1, False:0})

# 9. subCategory → collapse into a few broad types + fill NaN with mode
subcat_map = {
    'Daire':                   'Daire',
    'Apartment':               'Daire',
    'Villa':                   'Villa',
    'Yazlık':                  'Villa',
    'Bungalov':                'Villa',
    'Seaside&Summer House':    'Villa',
    'Müstakil Ev':             'House',
    'MüstakilEv':              'House',
    'M�stakil Ev':              'House',
    'Çiftlik Evi':             'House',
    'Köy Evi':                 'House',
    'Detached House':          'House',
    'Bina':                    'Building',
    'Prefabrik':               'Building',
    'Köşk':                    'Building',
    'Residence':               'Building',
    None:                      np.nan,   # we'll fill with mode next
}
data['subCategory'] = data['subCategory'].replace(subcat_map)
data['subCategory'].fillna(data['subCategory'].mode()[0], inplace=True)
```

```python
import numpy as np

# 1. ‘areas’ – only 1 missing: fill with the most common area
data['areas'].fillna(data['areas'].mode()[0], inplace=True)


# 2. Numeric columns – fill with median
for col in ['numberOfBuilding', 'entranceHeight', 'bedCount', 'age']:
    data[col].fillna(data[col].median(), inplace=True)


# 3. ‘heating’ & ‘fuel’ – treat missing as its own category
data['heating'] = data['heating'].fillna('Unknown')

# 4. ‘housingComplex’ – to much “None”
data.drop(columns=['housingComplex'], inplace=True)
data.drop(columns=['fuel'], inplace=True)




# 5. ‘authorizedRealtor’ – assume False where missing
data['authorizedRealtor'].fillna(False, inplace=True)


# 6. ‘registerState’ – duplicate of landRegisterName; drop it
data.drop(columns=['registerState'], inplace=True)
```
